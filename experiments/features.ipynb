```python
import pandas as pd
```

```python
#load dataset
accidents_df = pd.read_csv("../data/US_Accidents_March23_sampled_500k.csv")
```

```python
accidents_df['Weather_Condition'].value_counts()
```

```python
weather = accidents_df["Weather_Condition"].fillna("").str.lower()


accidents_df['has_rain'] = weather.str.contains(r"\brain\b|drizzle|drizziling|rainy",case=False,na=False).astype(int)
accidents_df['has_clouds'] = weather.str.contains(r"\bcloudy\b|cloud|clouds",case=False,na=False).astype(int)
accidents_df['has_snow'] = weather.str.contains("snow|snowy",case=False,na=False).astype(int)
accidents_df['is_sunny'] = weather.str.contains("clear",case=False,na=False).astype(int)
accidents_df['has_fog'] = weather.str.contains("fog|foggy",case=False,na=False).astype(int)

```
