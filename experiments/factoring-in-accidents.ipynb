```python
import pandas as pd
```

```python
# load dataset
car_accidents_df = pd.read_csv("../data/US_Accidents_March23_sampled_500k.csv")
car_accidents_df.columns
```

```python
# using the wind speeds below as counters
low_wind_speeds = 0 #represents 0-14 wind speeds (mph)
moderate_wind_speeds = 0 #represents 15-29 wind speeds (mph)
high_wind_speeds = 0 #represents 30+ wind speeds (mph)

for wind_mph in car_accidents_df['Wind_Speed(mph)']:
    if (wind_mph >= 0.0 and wind_mph < 15.0):
        low_wind_speeds += 1
    elif(wind_mph >=15.0 and wind_mph < 30.0):
        moderate_wind_speeds += 1
    else:
        high_wind_speeds += 1

print(f"Accidents with Wind Speeds from 0-14 mph: {low_wind_speeds}")
print(f"Accidents with Wind Speeds from 15-29 mph: {moderate_wind_speeds}")
print(f"Accidents with Wind Speeds from 30+ mph: {high_wind_speeds}")
```

```python
#making counters below to see if temperature plays a factor in car accidents
cold_temps = 0
hot_temps = 0
okay_temps = 0

for temp in car_accidents_df["Temperature(F)"]:
    if temp <= 40.0:        # for temps 40 and below (cold)
        cold_temps +=1
    elif temp >= 80.0: # for temps 80 and above (hot)
        hot_temps += 1
    else:                   # for temps between 41-79 (okay temps)
        okay_temps +=1

print(f"Accidents with temperatures that are below 40(F): {cold_temps}")
print(f"Accidents with temperatures that are above 80(F): {hot_temps}")
print(f"Accidents with temperatures that are between 41(F) - 79(F): {okay_temps}")
```

```python
visibility = car_accidents_df["Visibility(mi)"] 

low_visibility_count = (visibility < 1).sum() # dense fog, heavy rainfall, or heavy snowfall

reduced_visibility_count = ((visibility >= 1) & (visibility <=3)).sum() #reduced moderate rain, light snow, or fog

moderate_visibility_count = ((visibility >3) & (visibility <= 6)).sum() #moderate visbility, mist, light haze, or distant smoke.

clear_visibility_count = (visibility > 6).sum() # for visibility greater than 6 miles so clear

print(f"Accidents with visibility below 1 mile:  {low_visibility_count}")
print(f"Accidents with visibility between 1 mile and 3 miles:  {reduced_visibility_count}")
print(f"Accidents with visibility between 4 mile and 6 miles:  {moderate_visibility_count}")
print(f"Accidents with visibility between greater than 6 miles:  {clear_visibility_count}")
```

```python
# converting date and time strigns into datetime values to get sum
start_driving_timeframe = pd.to_datetime(car_accidents_df["Start_Time"],format='mixed')
end_driving_timeframe = pd.to_datetime(car_accidents_df["End_Time"],format='mixed')

#extracting only hour from start and end times
start_hours = start_driving_timeframe.dt.hour 
end_hours = end_driving_timeframe.dt.hour

# making a conditon for rush hour, and late night driving sum when accidents occured
morning_accidents = ((start_hours >=7) & (end_hours <=9)).sum()

late_night_accidents = ((start_hours >= 0 ) & (end_hours <=3)).sum()

#printing results
print(f"Accidents that occured during Rush Hour: {morning_accidents}")
print(f"Accidents that occured during Late-Night: {late_night_accidents}")

```
